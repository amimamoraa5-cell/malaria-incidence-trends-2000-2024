```python
metadata = pd.read_csv(
    "/kaggle/input/datasets/amimamoraa/malaria/Metadata_Country_API_SH.MLR.INCD.P3_DS2_en_csv_v2_572.csv"
)

metadata.head()
```

```python
print(metadata.columns.tolist())
```

```python
metadata["Region"].isnull().sum()
```

```python
metadata[metadata["Region"].isnull()][
    ["Country Code", "TableName", "Region", "IncomeGroup"]
].head(20)
```

```python
country_df = df.merge(
    metadata[["Country Code", "Region", "IncomeGroup"]],
    on="Country Code",
    how="left"
)

country_df = country_df[country_df["Region"].notna()].copy()

print("Country-level dataset:", country_df.shape)

```

```python
country_df[["Country Name", "Country Code", "Region", "IncomeGroup"]].head(10)
```

```python
country_df.isnull().sum()
```

```python
year_columns = [str(year) for year in range(2000, 2025)]

missing_percentage = (
    country_df[year_columns].isnull().mean() * 100
).round(2)

print(missing_percentage)
```

```python
country_df["available_years"] = (
    country_df[year_columns].notna().sum(axis=1)
)

country_df[
    ["Country Name", "available_years"]
].sort_values(
    by="available_years",
    ascending=False
).head(15)
```
